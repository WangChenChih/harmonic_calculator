# 設定空弦
因為泛音不可能可以比空弦音高還要低，因此在計算之前必須設定空弦。

變數```open_string_pitch```代表空弦音高之音名，而```open_string_octave```則指其位於之八度位置。舉例：中央C的音名為C4，因此其對應的```open_string_pitch```為字串```"C"```，並且```open_string_octave```為整數```4```。另外一個例子是二胡內弦，其空弦音高為D4，因此可將空弦設定為```open_string_pitch="D", open_string_octave=4```。雖然在這邊的設定是需要輸入音名，但其實不需拘泥，可以自行移調改為唱名，皆相通，只是要自行注意所有計算的符號定義。

````python
har = harmonic(open_string_pitch="D", open_string_octave=4)
````

# 計算人工泛音
以二胡為例，人工泛音的產生方式通常為一指（食指 ```forefinger```）按實音，而用四指或三指（在這裡我稱為第二指```second```）虛按某特定處。函數

```calculate_artificial_harmonic(pitch_name_goal, octave_goal)```

將輸出食指按弦處之實音音高與（```forefinger: ```）第二指按弦處之實音音高（```second: ```）。其中```pitch_name_goal```為欲產生之人工泛音之音名，而```octave_goal```為其所在之八度位置。


````python
har.calculate_artificial_harmonic(pitch_name_goal="E", octave_goal=6)
````
